This notebook checks the source and the common population before the owners deepen their axes. It exports no individual respondent rows. The project uses 2026 only; 2025 comparisons require separate harmonisation.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if not (ROOT / "src").is_dir():
    raise RuntimeError("Open this notebook from the project root or its notebooks folder.")
sys.path.insert(0, str(ROOT))
from src.data import ensure_data, load_responses, load_schema
from src.population import prepare_population

ensure_data()
population = prepare_population(load_responses())
schema = load_schema()
print("2026 · France · coding as primary job · eligible respondents:", len(population))
```

```python
raw = load_responses()
print("Source shape:", raw.shape)
print("Unique response IDs:", raw["ResponseId"].is_unique)
print("French respondents:", raw["Country"].eq("France").sum())
display(raw.loc[raw["Country"].eq("France"), "CodeMain"].value_counts(dropna=False).to_frame("respondents"))
```

```python
fields = ["AISelect", "AITrust", "AICodeAgentHaveWorkedWith", "AIWorkCurrentlyUse", "AIWorkNext", "LearnAI", "LearnAIHow", "AIAdopt"]
display(schema.loc[schema["column"].isin(fields), ["column", "question", "sub", "transformation"]])
display(population[fields].notna().sum().rename("answered out of 628").to_frame())
display(population["AITrust"].value_counts(dropna=False).rename("respondents").to_frame())
```
